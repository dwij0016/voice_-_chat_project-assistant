```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
```

```python
a = pd.read_csv("C:\\unknown\\voice_&_chat_project assistant\\data\\intents.csv")
print(a)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    a.text, a.label, test_size = 0.2, stratify = a.label, random_state = 42)
```

```python
clf = make_pipeline(
    TfidfVectorizer(ngram_range=(1, 2), lowercase=True),
    LogisticRegression(max_iter=1000),
)
```

```python
clf = make_pipeline(
    TfidfVectorizer(ngram_range=(1, 2), lowercase=True),
    LogisticRegression(max_iter=1000),
)
```

```python
clf.fit(x_train, y_train)
```

```python
print(classification_report(y_test, clf.predict(x_test)))
ConfusionMatrixDisplay.from_estimator(clf, x_test, y_test, xticks_rotation=42)
```

```python
joblib.dump(clf, "../models/intent.joblib")
```
